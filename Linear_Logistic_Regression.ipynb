```python
import numpy as np
import pandas as pd
```

```python
data = {
    'Hours': [2, 3, 4, 5, 6],
    'Marks': [25, 35, 45, 55, 65]
}

df = pd.DataFrame(data)

X = df['Hours'].values
y = df['Marks'].values
```

```python
class LinearRegression:
    def fit(self, X, y):
        self.m = np.sum((X - X.mean()) * (y - y.mean())) / np.sum((X - X.mean()) ** 2)
        self.c = y.mean() - self.m * X.mean()

    def predict(self, X):
        return self.m * X + self.c
```

```python
model = LinearRegression()
model.fit(X, y)

predictions = model.predict(X)
```

```python
print("Actual Marks:", y)
print("Predicted Marks:", predictions)
```

```python
data = {
    'Hours': [2, 3, 4, 5, 6, 7],
    'Result': [0, 0, 0, 1, 1, 1]
}

df = pd.DataFrame(data)

X = df['Hours'].values
y = df['Result'].values
```

```python
class LogisticRegression:
    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def fit(self, X, y):
        self.m = 0
        self.c = 0
        lr = 0.01

        for i in range(1000):
            z = self.m * X + self.c
            y_pred = self.sigmoid(z)

            dm = np.mean((y_pred - y) * X)
            dc = np.mean(y_pred - y)

            self.m = self.m - lr * dm
            self.c = self.c - lr * dc

    def predict(self, X):
        z = self.m * X + self.c
        probability = self.sigmoid(z)
        return (probability >= 0.5).astype(int)
```

```python
model = LogisticRegression()
model.fit(X, y)

predictions = model.predict(X)
```

```python
print("Actual Results:", y)
print("Predicted Results:", predictions)
```
